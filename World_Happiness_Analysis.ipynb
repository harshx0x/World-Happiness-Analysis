```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

print("All libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()

```

```python
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("Total Rows:", df.shape[0])

print("Total Columns:", df.shape[1])
```

```python
print("Columns in our dataset:")

print(df.columns)
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df_clean = df.copy()

print("Dataset copied successfully!")
```

```python
duplicates = df_clean.duplicated().sum()

print("Total duplicate rows:", duplicates)
```

```python
df_clean = df_clean.drop_duplicates()

print("Duplicate records removed successfully!")
```

```python
print(df_clean.isnull().sum())
```

```python
df_clean.columns = [
    'Rank',
    'Country',
    'Happiness_Score',
    'GDP',
    'Social_Support',
    'Life_Expectancy',
    'Freedom',
    'Generosity',
    'Corruption'
]

print(df_clean.columns)
```

```python
df_clean.head()
```

```python
top_10 = df_clean.sort_values(
    by='Happiness_Score',
    ascending=False
).head(10)

print("Top 10 Happiest Countries in 2019")

print(top_10[['Country', 'Happiness_Score']])
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10,
    x='Happiness_Score',
    y='Country'
)


plt.title('Top 10 Happiest Countries in 2019',
          fontsize=16)

plt.xlabel('Happiness Score')

plt.ylabel('Country')

plt.show()
```

```python
bottom_10 = df_clean.sort_values(
    by='Happiness_Score',
    ascending=True
).head(10)

print("10 Least Happy Countries in 2019")

print(bottom_10[['Country', 'Happiness_Score']])
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=bottom_10,
    x='Happiness_Score',
    y='Country'
)

plt.title('10 Least Happy Countries in 2019',
          fontsize=16)

plt.xlabel('Happiness Score')

plt.ylabel('Country')

plt.show()
```

```python
average_happiness = df_clean['Happiness_Score'].mean()

print("Average Happiness Score:",
      round(average_happiness, 2))
```

```python
highest = df_clean['Happiness_Score'].max()

lowest = df_clean['Happiness_Score'].min()

print("Highest Happiness Score:", highest)

print("Lowest Happiness Score:", lowest)
```
